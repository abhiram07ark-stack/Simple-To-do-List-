```python
todo_list = []

while True:
    print("\n1. Add to list")
    print("2. View list")
    print("3. Remove from list")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        task = input("Enter a task: ")
        todo_list.append(task)
        print("Task added")

    elif choice == "2":
        if len(todo_list) == 0:
            print("No tasks found.")
        else:
            print("Your tasks:")
            todo_list = ["Homework", "Study", "Exercise"]
len(todo_list)  # → 3


    elif choice == "3":
        if len(todo_list) == 0:
            print("No tasks to remove.")
        else:
            num = int(input("Enter task number to remove: "))
            todo_list.pop(num - 1)
            print("Task removed.")

    elif choice == "4":
        print("Exiting program.")
        break

    else:
        print("Invalid choice. Try again.")
```

What I Learned: Simple To-Do List in Python

In this program, I learned how to use a list to store multiple tasks and modify it during program execution. I used a while loop to repeatedly show a menu until the user chooses to exit. The user’s input was handled using if-elif-else statements to perform different actions such as adding, viewing, and removing tasks.

I learned that input() always returns a string, so choices must be compared with string values. I also learned how len() is used to find the number of items in a list, and how range(len(list)) allows looping through list indexes. Since list indexes start at 0, I used i + 1 when displaying tasks to make numbering user-friendly.
